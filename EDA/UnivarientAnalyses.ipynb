Univariate vs Bivariate vs Multivariate

Ye ML/EDA me bahut important distinction hai:

Univariate
     ↓
1 variable
Age

Bivariate
     ↓
2 variables
Age + Fare

Multivariate
     ↓
3+ variables
Age + Fare + Pclass + Survived

Example:

# Univariate
sns.histplot(df['Age'])

# Bivariate
sns.scatterplot(data=df, x='Age', y='Fare')

# Multivariate
sns.scatterplot(
    data=df,
    x='Age',
    y='Fare',
    hue='Survived'
)

Simple rule yaad rakho:

Univariate = "Is ek column ke andar data kaisa hai?"

EDA karte waqt usually pehle Univariate → Bivariate → Multivariate analysis karna useful hota hai.

Boxplot

1. Boxplot hota kya hai?

Maan lo Titanic dataset mein:

df["Age"]

mein passengers ki ages hain.

Agar tum:

sns.boxplot(data=df, x="Age")
plt.show()

chalaoge, to ek box aur uske dono taraf lines/dots dikhengi.

Boxplot basically data ko 5 important values ke through summarize karta hai:

Minimum ─── Q1 ┃──── Median ────┃ Q3 ─── Maximum
                  BOX

Lekin yahan ek important point hai: boxplot ka "minimum" aur "maximum" usually actual absolute min/max nahi, balki non-outlier range ke endpoints hote hain.

2. Boxplot ke 5 concepts

###① Q1 — First Quartile

Data ko ascending order mein lagao.

Q1 = 25% data isse neeche hota hai.

###② Median — Q2

50% data neeche, 50% upar.

Ye box ke andar middle line hoti hai.

###③ Q3 — Third Quartile

75% data isse neeche hota hai.

###④ IQR

Ye bahut important hai:

IQR = Q3 - Q1

IQR batata hai ki middle 50% data kitna spread hai.

###⑤ Outliers

Jo values normal range se bahut door hain, unhe boxplot alag dots ke form mein dikha sakta hai.

3. Visual structure

Suppose Age ka boxplot kuch aisa hai:

       ────────────────
       Q1     Median    Q3
        │       │       │
────────┤┌─────────────┐├────────
        ││             ││
        ││    BOX      ││
────────┤└─────────────┘├────────
                         •
                         •
                         •
                      Outliers

Box ka left edge = Q1

Box ke andar line = Median

Box ka right edge = Q3

Box ke bahar whiskers = non-outlier values ki range

Alag dots = potential outliers

4. Outlier actually kaise decide hota hai?

Yahan IQR ka use hota hai.

IQR = Q3 - Q1

Phir:

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Jo value:

< Lower Bound

ya

> Upper Bound

hai, use boxplot potential outlier ke roop mein show karta hai.

Example

Maan lo:

Q1 = 20
Q3 = 40

To:

IQR = 40 - 20
    = 20

Lower bound:

20 - 1.5(20)
= -10

Upper bound:

40 + 1.5(20)
= 70

To 70 se upar ki ages potential outliers hongi.

Sabse important use:

Outlier detection

Suppose tumhare dataset mein salary hai:

20k
25k
30k
35k
40k
45k
50 lakh

50 lakh baaki values se bahut alag hai.

Boxplot tumhe immediately signal de sakta hai ki:

50 lakh → potential outlier

Phir ML preprocessing mein tum investigate karte ho:

Ye genuine value hai ya data-entry mistake?

Uske baad decide karte ho:

remove karna hai
cap/winsorize karna hai
transform karna hai
ya waise hi rakhna hai

Important: Boxplot mein outlier dikhna automatically ye nahi bolta ki value galat hai.

Ek line mein yaad rakho:

Histogram/Displot → data ka distribution kaisa hai
Boxplot → data ka spread + median + quartiles + potential outliers

Aur Boxplot ko properly samajhne ke liye Quartile (Q1, Median, Q3) + IQR sabse important concepts hain.
