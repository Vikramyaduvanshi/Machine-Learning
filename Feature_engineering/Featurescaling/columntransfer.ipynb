```python
import pandas as pd
import numpy as np
```

```python
df= pd.read_csv("d:/Machine_learning/data/covid_toy.csv")
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(df.drop(columns=['has_covid']),df['has_covid'],test_size=.2)
```

```python
x_train
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
```

```python
transformer = ColumnTransformer(transformers=[
    ('tf1', SimpleImputer(), ['fever']),
    ('tf2', OrdinalEncoder(categories=[["Mild", "Strong"]]), ['cough']),
    ('tf3', OneHotEncoder(sparse_output=False, drop='first'), ['gender', 'city'])
],remainder='passthrough')
```

```python
transformer.fit_transform(x_train).shape
```

```python
transformer.transform(x_test).shape
```
