1. Normalization kya hoti hai?

Normalization ek feature scaling technique hai, jisme hum numerical values ko ek fixed range mein convert karte hain.

MinMaxScaler ka default range hota hai:

$$ [0,1] $$

Matlab original data ki minimum value 0 aur maximum value 1 ban jaati hai.

Example

Maan lo tumhare paas Age feature hai:

Age
20
30
40
50
60

MinMaxScaler lagane ke baad:

Original Age	Scaled Age
20	0.00
30	0.25
40	0.50
50	0.75
60	1.00

Notice karo, values ka order same hai, lekin ab saari values 0 se 1 ke beech aa gayi hain.

2. MinMaxScaler ka mathematical formula

Ye formula sabse important hai:

$$ \boxed{X_{scaled}=\frac{X-X_{min}}{X_{max}-X_{min}}} $$

Har term ka meaning:

Term	Meaning
\(X\)	Current/original value
\(X_{min}\)	Feature ki minimum value
\(X_{max}\)	Feature ki maximum value
\(X_{scaled}\)	Scaled value

Logic: Original value mein se minimum subtract karte hain, phir maximum aur minimum ke difference se divide karte hain.

Isse minimum value 0 aur maximum value 1 ban jaati hai.

3. Manually calculation karte hain

Maan lo:

Age = [20, 30, 40, 50, 60]

Sabse pehle:

Xmin = 20
Xmax = 60

Ab formula:

$$ X_{scaled}=\frac{X-20}{60-20} $$

Denominator:

$$ 60-20=40 $$

Ab har value ko individually calculate karenge.

Case 1: Age = 20
$$ \frac{20-20}{40}=\frac{0}{40}=0 $$
Case 2: Age = 30
$$ \frac{30-20}{40}=\frac{10}{40}=0.25 $$
Case 3: Age = 40
$$ \frac{40-20}{40}=\frac{20}{40}=0.50 $$
Case 4: Age = 50
$$ \frac{50-20}{40}=\frac{30}{40}=0.75 $$
Case 5: Age = 60
$$ \frac{60-20}{40}=\frac{40}{40}=1 $$

Final result:

Original Age  →  Scaled Age

20            →  0.00
30            →  0.25
40            →  0.50
50            →  0.75
60            →  1.00

Yahi MinMaxScaler internally karta hai!

4. Python mein MinMaxScaler

Ab sklearn se same calculation karte hain.

import pandas as pd
from sklearn.preprocessing import MinMaxScaler

df = pd.DataFrame({
    "Age": [20, 30, 40, 50, 60]
})

scaler = MinMaxScaler()

df["Age_scaled"] = scaler.fit_transform(df[["Age"]])

print(df)

Output:

   Age  Age_scaled
0   20        0.00
1   30        0.25
2   40        0.50
3   50        0.75
4   60        1.00

df[["Age"]]

normally ek Pandas Series deta hai, jabki:

ek single-column DataFrame deta hai.

6. MinMaxScaler aur StandardScaler mein difference
StandardScaler	MinMaxScaler
Mean approximately 0 karta hai	Minimum ko 0 karta hai (default range mein)
Standard deviation approximately 1 karta hai	Maximum ko 1 karta hai (default range mein)
Fixed range guarantee nahi karta	Training data ko fixed range mein map karta hai
Formula: (X - mean) / std	Formula: (X - min) / (max - min)
Ek important ML insight

MinMaxScaler outliers se kaafi affect ho sakta hai.

Example:

Age = [20, 30, 40, 50, 200]

Yahan 200 ek extreme value hai. MinMaxScaler isko maximum maan kar baaki values ko 0 ke paas compress kar dega.

Isliye scaling method choose karte waqt data distribution aur outliers ko samajhna useful hota hai.

7. Train-Test Split ke saath MinMaxScaler

Jaise tumne StandardScaler mein seekha tha, yahan bhi same golden rule hai:

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler

X = df[["Age", "EstimatedSalary"]]
y = df["Purchased"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=0
)

scaler = MinMaxScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
fit_transform(X_train) → training data ka min/max seekho aur scale karo.
transform(X_test) → training data ke min/max se test data scale karo.

Test data par dobara fit() nahi karna, warna data leakage ho sakta hai.

Quick revision 🧠
MinMaxScaler
     ↓
Minimum aur Maximum find karo
     ↓
(X - Xmin) / (Xmax - Xmin)
     ↓
Minimum → 0
Maximum → 1
     ↓
Baaki values bhi [0,1] range mein

Ek baat yaad rakhna: default MinMaxScaler training data ko [0,1] range mein map karta hai. Agar unseen test data training ke min/max se bahar ho, to uski scaled value 0 se kam ya 1 se zyada bhi ho sakti hai.
