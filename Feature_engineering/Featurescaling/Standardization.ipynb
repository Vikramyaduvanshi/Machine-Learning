5. StandardScaler exactly yehi formula use karta hai

Python mein manually formula calculate karne ki zarurat nahi.

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

StandardScaler internally roughly ye karta hai:

1. Mean calculate karo
       ↓
2. Standard deviation calculate karo
       ↓
3. Har value par
   (x - mean) / std
       ↓
4. Scaled data return karo
6. fit() ka exact meaning

7. transform() kya karta hai?
scaler.transform(X)

Ab jo mean aur standard deviation fit() ke time seekhe the, unko use karke values ko transform karta hai.

So:

scaler.fit(X_train)

= seekho

scaler.transform(X_train)

= seekhe hue parameters se convert karo

8. fit_transform() kya hai?

Ye dono ka shortcut hai:

scaler.fit_transform(X)

Equivalent conceptually:

scaler.fit(X)
X_scaled = scaler.transform(X)

Isliye beginners ko ye confusing lagta hai, but simple hai:

fit = learn
transform = apply
fit_transform = learn + apply

9. ML mein sabse important rule ⚠️

Suppose:

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2
)

Ab:

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
Aisa kyun?

Because test data se kuch seekhna nahi hai.

❌ Wrong:

X_test = scaler.fit_transform(X_test)

Yahan test data ka mean/std bhi calculate ho jayega.

✅ Correct:

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

Training data:

fit → mean/std seekho
transform → scale karo

Test data:

sirf transform

Ye data leakage prevent karta hai.

10. Ek line mein Standardization

Yaad rakhna:

Standardization = har feature ko uske mean se center karke standard deviation se divide karna, jisse feature approximately mean 0 aur std 1 scale par aa jaye.

Formula:

z = (x - mean) / standard deviation

Aur sklearn:

StandardScaler()

Standardization
      ↓

      ↓
μ = Mean
σ = Standard Deviation
      ↓
StandardScaler
      ↓
fit → learn mean/std
transform → apply
fit_transform → learn + apply
